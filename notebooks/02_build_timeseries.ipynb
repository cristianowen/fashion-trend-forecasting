```python
import duckdb
import pandas as pd

result = duckdb.sql("""
    SELECT COUNT(*) as total_rows, 
           MIN(t_dat) as earliest_date, 
           MAX(t_dat) as latest_date
    FROM '../data/raw/transactions_train.csv'
""").df()

result
```

```python
trousers_ts = duckdb.sql("""
    SELECT 
        t.t_dat AS date,
        tr.fit_category,
        COUNT(*) AS units_sold
    FROM '../data/raw/transactions_train.csv' t
    JOIN '../data/interim/trousers_tagged.csv' tr
        ON t.article_id = tr.article_id
    GROUP BY t.t_dat, tr.fit_category
    ORDER BY t.t_dat
""").df()

print(trousers_ts.shape)
trousers_ts.head(10)
```

```python
jackets_ts = duckdb.sql("""
    SELECT 
        t.t_dat AS date,
        j.style_category,
        COUNT(*) AS units_sold
    FROM '../data/raw/transactions_train.csv' t
    JOIN '../data/interim/jackets_tagged.csv' j
        ON t.article_id = j.article_id
    GROUP BY t.t_dat, j.style_category
    ORDER BY t.t_dat
""").df()

print(jackets_ts.shape)
jackets_ts.head(10)
```

```python
trousers_ts.to_csv('../data/processed/trousers_daily_sales.csv', index=False)
jackets_ts.to_csv('../data/processed/jackets_daily_sales.csv', index=False)
print("Saved both time series to data/processed/")
```
