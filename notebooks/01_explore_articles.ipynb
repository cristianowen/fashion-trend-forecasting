```python
import pandas as pd

articles = pd.read_csv('../data/raw/articles.csv')
print(articles.shape)
articles.head()
```

```python
articles['product_type_name'].value_counts().head(30)
```

```python
trousers = articles[articles['product_type_name'] == 'Trousers']
pd.set_option('display.max_colwidth', None)
trousers[['prod_name', 'detail_desc']].sample(15, random_state=1)
```

```python
trousers_desc = trousers['detail_desc'].fillna('').str.lower()

keywords = ['skinny', 'slim', 'straight', 'tapered', 'wide', 'boyfriend', 
            'flare', 'flared', 'baggy', 'bootcut', 'relaxed', 'loose', 
            'super-skinny', 'regular']

for kw in keywords:
    count = trousers_desc.str.contains(kw).sum()
    print(f"{kw:15s}: {count}")
```

```python
jackets = articles[articles['product_type_name'] == 'Jacket']
jackets_desc = jackets['detail_desc'].fillna('').str.lower()

jacket_keywords = ['denim', 'leather', 'canvas', 'bomber', 'puffer', 
                    'parka', 'biker', 'fleece', 'quilted', 'padded',
                    'corduroy', 'suede', 'wool', 'utility']

for kw in jacket_keywords:
    count = jackets_desc.str.contains(kw).sum()
    print(f"{kw:12s}: {count}")

print(f"\nTotal jackets: {len(jackets)}")
```

```python
jackets_desc = jackets['detail_desc'].fillna('').str.lower()

leather_biker = jackets_desc.str.contains('leather|biker').sum()
print(f"Leather + Biker combined: {leather_biker}")
print(f"Denim: {jackets_desc.str.contains('denim').sum()}")
print(f"Bomber: {jackets_desc.str.contains('bomber').sum()}")
```

```python
def tag_trouser_fit(desc):
    desc = str(desc).lower()
    if 'skinny' in desc or 'slim' in desc:
        return 'skinny_slim'
    elif 'tapered' in desc:
        return 'tapered'
    elif 'straight' in desc:
        return 'straight'
    elif 'wide' in desc or 'relaxed' in desc or 'loose' in desc:
        return 'wide_relaxed'
    elif 'regular' in desc:
        return 'regular'
    else:
        return 'other'

trousers = trousers.copy()
trousers['fit_category'] = trousers['detail_desc'].apply(tag_trouser_fit)
trousers['fit_category'].value_counts()
```

```python
def tag_jacket_style(desc):
    desc = str(desc).lower()
    if 'denim' in desc:
        return 'denim'
    elif 'leather' in desc or 'biker' in desc:
        return 'leather_biker'
    elif 'bomber' in desc:
        return 'bomber'
    else:
        return 'other'

jackets = jackets.copy()
jackets['style_category'] = jackets['detail_desc'].apply(tag_jacket_style)
jackets['style_category'].value_counts()
```

```python
trousers_tagged = trousers[trousers['fit_category'] != 'other'][['article_id', 'product_code', 'prod_name', 'fit_category']]
jackets_tagged = jackets[jackets['style_category'] != 'other'][['article_id', 'product_code', 'prod_name', 'style_category']]

trousers_tagged.to_csv('../data/interim/trousers_tagged.csv', index=False)
jackets_tagged.to_csv('../data/interim/jackets_tagged.csv', index=False)

print(f"Saved {len(trousers_tagged)} trousers and {len(jackets_tagged)} jackets")
```
